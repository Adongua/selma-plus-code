```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 从CSV文件中读取数据
data = pd.read_csv('rt-relu.csv')
rt_torch = data["rt_plain"]
rt_enc_last = data["rt_enc_last"]
rt_enc_now = data["rt_enc_now"]
```

```python
x = list(range(1,9))

plt.figure(figsize=(5.9, 4.5))  # 保证图片尺寸一致

plt.plot(x, rt_torch, label='PyTorch', marker='o', color="darkslategrey", linewidth=2.5, markersize=7, markerfacecolor='white', markeredgewidth=2.5)
plt.plot(x, rt_enc_last, label='CrypTen', marker='s', color="darkseagreen", linewidth=2.5, markersize=7, markerfacecolor='white', markeredgewidth=2.5)
plt.plot(x, rt_enc_now, label='SELMA', marker='^', color="seagreen", linewidth=2.5, markersize=7, markerfacecolor='white', markeredgewidth=2.5)

# 添加图例，调大字体
plt.legend(fontsize=15)

# 添加标题和标签，调大字体
plt.xlabel('Number of Parties', fontsize=15)
plt.ylabel('Runtime(s)', fontsize=15)

# 坐标轴刻度字体变大
plt.tick_params(axis='both', labelsize=13)

# 添加网格
plt.grid(True, linestyle='--')

# 保存
plt.savefig('rt-relu-gpu.png', dpi=300, bbox_inches='tight')  # 加上bbox_inches参数

# 显示图像
plt.show()
```

```python
# 从CSV文件中读取数据
data = pd.read_csv('bs-rt-relu.csv')
rt_torch = data["rt_plain"]
rt_enc_last = data["rt_enc_last"]
rt_enc_now = data["rt_enc_now"]
```

```python
x = [1,2,4,8,16,32,64,128]

# 创建图形，调整尺寸比例
plt.figure(figsize=(5.9, 4.5))  # 调小了图形尺寸

# 绘制曲线
plt.semilogx(x, rt_torch, label='PyTorch', marker='o', color="darkslategrey", linewidth=2.5, markersize=7, markerfacecolor='white', markeredgewidth=2.5)
plt.semilogx(x, rt_enc_last, label='CrypTen', marker='s', color="darkseagreen", linewidth=2.5, markersize=7, markerfacecolor='white', markeredgewidth=2.5)
plt.semilogx(x, rt_enc_now, label='SELMA', marker='^', color="seagreen", linewidth=2.5, markersize=7, markerfacecolor='white', markeredgewidth=2.5)


# 获取当前轴对象
ax = plt.gca()

# 隐藏次刻度
ax.minorticks_off()  # 这行替代了之前报错的代码

# 设置主刻度
plt.xticks(x, [str(i) for i in x])


# 设置网格（虚线样式）
plt.grid(True, linestyle='--')

# 添加标签
plt.xlabel('Batch Size', fontsize=15)
plt.ylabel('Runtime(s)', fontsize=15)

# 坐标轴刻度字体变大
plt.tick_params(axis='both', labelsize=13)

# 添加图例，调大字体
plt.legend(fontsize=15)

# 调整布局
plt.tight_layout()

# 保存图像
plt.savefig('bs-rt-relu-gpu.png', dpi=300, bbox_inches='tight')

# 显示图像
plt.show()
```
