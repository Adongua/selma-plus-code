```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 从CSV文件中读取数据
data = pd.read_csv('single_party/func_benchmarks.csv')
# print(data)
# 提取需要的列数据
functions = data['function']
runtime_1 = data['runtime'] 
runtime_crypten_1 = data['runtime crypten']
# print(runtime)
# print(runtime_crypten)

# 绘制柱形图
bar_width = 0.35
index = range(len(functions))

plt.bar(index, runtime_1, bar_width, label='runtime pytorch')
plt.bar([i + bar_width for i in index], runtime_crypten_1, bar_width, label='runtime crypten')

# 设置横坐标刻度和标签
plt.xticks([i + bar_width/2 for i in index], functions, rotation='vertical')

# 添加图例
plt.legend()

# 添加标题和标签
plt.title('Runtime Comparison')
plt.xlabel('functions')
plt.ylabel('Runtime(s)')

# 保存
plt.savefig('img/Runtime Comparison in cpu.png')

# 显示图形
plt.show()
```

```python
# 从CSV文件中读取数据
data = pd.read_csv('multi_parties/2/func_benchmarks.csv')
runtime = []
runtime_crypten = []
runtime.append(runtime_1)
runtime_crypten.append(runtime_crypten_1)
for i in range(2,9):
    path = 'multi_parties/'+ str(i) +'/func_benchmarks.csv'
    data = pd.read_csv(path)
    runtime.append(data['runtime'])
    runtime_crypten.append(data['runtime crypten'])
```

```python
x = list(range(1,9))
for i in range(0,len(functions)): 
    y = []
    for ele in runtime:
        y.append(ele[i])
    plt.plot(x,y,label='runtime pytorch '+functions[i])
    
    y_crypten = []
    for ele in runtime_crypten:
        y_crypten.append(ele[i])
    plt.plot(x,y_crypten,label='runtime crypten '+functions[i])
    
    # 添加图例
    plt.legend()

    # 添加标题和标签
    plt.title('Runtime Comparison Trend')
    plt.xlabel('Parties Number')
    plt.ylabel('Runtime(s)')
    
    # 保存
    plt.savefig('img/Runtime Comparison Trend for ' + functions[i] + ' in cpu.png')
    
    plt.show()
```

```python
x = list(range(1,9))
for i in range(0,len(functions)): 
    y = []
    for ele in runtime:
        val = f'{ele[i]:e}'
        val = val.split('e')[1]
        y.append('1e'+val)
    plt.plot(x,y,label='runtime pytorch '+functions[i])
    
    y_crypten = []
    for ele in runtime_crypten:
        val = f'{ele[i]:e}'
        val = val.split('e')[1]
        y_crypten.append('1e'+val)
    plt.plot(x,y_crypten,label='runtime crypten '+functions[i])
    
    # 添加图例
    plt.legend()

    # 添加标题和标签
    plt.title('Runtime Comparison Trend')
    plt.xlabel('Parties Number')
    plt.ylabel('Runtime')
    
    # 保存
    plt.savefig('img_e/Runtime Comparison Trend for ' + functions[i] + ' in cpu.png')
    
    plt.show()
```
